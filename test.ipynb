```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
netflix_df = pd.read_csv("netflix_data.csv")
netflix_df.head()
```

```python
movies=netflix_df["type"].value_counts()        #Count movies vs TV shows
gen=netflix_df['genre'].value_counts()          #Top genres/categories
yr=netflix_df["release_year"].value_counts()    #Distribution of release years
conty=netflix_df["country"].value_counts()      #producing countries
```

```python
#show genre in bar chart
gen.plot(kind="bar", stacked=True, figsize=(10,5))
plt.title("Number of Movies per Genre")
plt.xlabel("Genre")
plt.ylabel("count")
plt.show()
```

```python
cont=conty.head(10)
cont.plot(kind='barh', figsize=(8,5), color='lightgreen')
plt.title("Top 10 Countries Producing Content")
plt.xlabel("Count")
plt.ylabel("Country")
plt.show()
```

```python
gen.plot(kind='pie', autopct='%1.1f%%', figsize=(4,4))
plt.title("percentage of Movies per Genre")
plt.ylabel("")  # remove y-label for cleaner look
plt.show()
```

```python
#how many movies fall into each combination
country_genre_table=pd.crosstab(netflix_df["country"], netflix_df["genre"])
country_genre_table
```

```python
# Top 10 countries by Drama movies
drama_counts = country_genre_table['Dramas'].sort_values(ascending=False).head(10)

drama_counts.plot(kind='bar', figsize=(8,5), color='purple')
plt.title("Top 10 Countries Producing Drama Movies")
plt.xlabel("Country")
plt.ylabel("Count")
plt.show()
```
